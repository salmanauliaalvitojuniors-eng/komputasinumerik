## Ide dasar metode Bisection

1. Cari dua titik **a** dan **b** sehingga f(a) dan f(b) **berlawanan tanda** (satu positif, satu negatif). Artinya grafik pasti memotong sumbu x di antara a dan b.
2. Hitung titik tengah: **c = (a + b) / 2**.
3. Lihat tanda f(c):
   - Kalau f(a) dan f(c) berlawanan tanda, akar ada di [a, c], jadi b diganti c.
   - Kalau tidak, akar ada di [c, b], jadi a diganti c.
4. Ulangi sampai selang cukup kecil (lebih kecil dari toleransi) atau f(c) hampir nol.

Setiap iterasi membuat selang menyempit jadi setengahnya.

```python
# ===== CELL 1: IMPORT LIBRARY =====
# math      : untuk fungsi exp() dan log()
# pandas    : untuk menampilkan tabel iterasi yang rapi
# matplotlib: untuk menggambar grafik fungsi
import math
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ===== CELL 2: DEFINISI FUNGSI f(x) =====
# Fungsi yang akan dicari akarnya: f(x) = e^x - 3x^2
def f(x):
    # math.exp(x) artinya e pangkat x
    # 3 * x**2 artinya 3 dikali x kuadrat
    return math.exp(x) - 3 * x**2

# Uji cepat: lihat nilai f di beberapa titik
print("f(0) =", f(0))
print("f(1) =", f(1))
```

```python
# ===== CELL 3: PARAMETER AWAL =====
a = 0.0            # batas kiri selang
b = 1.0            # batas kanan selang
toleransi = 1e-6   # selang dianggap cukup kecil kalau (b - a) lebih kecil dari ini
maks_iterasi = 50  # batas jumlah iterasi supaya program tidak berjalan selamanya

# Syarat metode Bisection: f(a) dan f(b) harus berlawanan tanda.
# Kalau f(a) * f(b) < 0, berarti tandanya berlawanan.
if f(a) * f(b) >= 0:
    print("Selang tidak valid: f(a) dan f(b) harus berlawanan tanda.")
else:
    print("Selang valid, akar berada di antara", a, "dan", b)
```

```python
# ===== CELL 4: PROSES BISECTION =====
# Daftar kosong untuk menyimpan hasil tiap iterasi (dipakai untuk tabel)
hasil = []

for i in range(1, maks_iterasi + 1):
    c = (a + b) / 2        # titik tengah selang
    fc = f(c)              # nilai fungsi di titik tengah

    # simpan data iterasi ke dalam daftar
    hasil.append({"iterasi": i, "a": a, "b": b, "c": c, "f(c)": fc, "lebar": b - a})

    # kondisi berhenti: f(c) hampir nol, atau selang sudah cukup kecil
    if abs(fc) < toleransi or (b - a) / 2 < toleransi:
        break

    # tentukan sisi mana yang berisi akar
    if f(a) * fc < 0:
        b = c   # akar di sisi kiri, geser batas kanan ke c
    else:
        a = c   # akar di sisi kanan, geser batas kiri ke c

akar = c
print("Proses selesai setelah", i, "iterasi")
```

```python
# ===== CELL 5: TABEL ITERASI =====
# Ubah daftar hasil menjadi tabel pandas supaya mudah dibaca
tabel = pd.DataFrame(hasil)
tabel
```

```python
# ===== CELL 6: HASIL AKHIR DAN VERIFIKASI =====
print("Akar hampiran  x =", round(akar, 6))
print("Nilai f(x)       =", f(akar))   # harus sangat dekat dengan 0
print("Jumlah iterasi   =", len(hasil))

# Verifikasi: sisi kiri (e^x) dan sisi kanan (3x^2) harus hampir sama
print("e^x  =", math.exp(akar))
print("3x^2 =", 3 * akar**2)
```

```python
# ===== CELL 7: GRAFIK =====
# Buat 200 titik x dari 0 sampai 1.2 untuk menggambar kurva
xs = [i * 1.2 / 200 for i in range(201)]
ys = [f(x) for x in xs]

plt.plot(xs, ys, label="f(x) = e^x - 3x^2")
plt.axhline(0, color="black", linewidth=0.8)             # garis sumbu x
plt.scatter([akar], [f(akar)], color="red", zorder=5,
            label="akar = " + str(round(akar, 4)))        # titik akar
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Metode Bisection")
plt.legend()
plt.grid(True)
plt.show()
```
