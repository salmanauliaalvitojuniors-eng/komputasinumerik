```python
# ===== CELL 1: INPUT KOEFISIEN =====
# Sistem persamaan linear 3 variabel berbentuk:
#   a1*x + b1*y + c1*z = d1
#   a2*x + b2*y + c2*z = d2
#   a3*x + b3*y + c3*z = d3

# Persamaan 1: 2x + 1y - 1z = 8
# a = koefisien x, b = koefisien y, c = koefisien z, d = konstanta (sisi kanan)
a1, b1, c1, d1 = 2, 1, -1, 8

# Persamaan 2: -3x - 1y + 2z = -11
a2, b2, c2, d2 = -3, -1, 2, -11

# Persamaan 3: -2x + 1y + 2z = -3
a3, b3, c3, d3 = -2, 1, 2, -3

# Menampilkan ketiga persamaan agar bisa diperiksa apakah input sudah benar
print(f"({a1})x + ({b1})y + ({c1})z = {d1}")
print(f"({a2})x + ({b2})y + ({c2})z = {d2}")
print(f"({a3})x + ({b3})y + ({c3})z = {d3}")
```

```python
# ===== CELL 2: MEMASTIKAN KOEFISIEN x PADA PERSAMAAN 1 TIDAK NOL =====
# Pada metode substitusi, x dinyatakan dari persamaan 1 dengan cara membagi dengan a1.
# Jika a1 = 0, pembagian tidak mungkin dilakukan, sehingga urutan persamaan harus ditukar.

if a1 == 0:                              # cek apakah a1 bernilai nol
    if a2 != 0:                          # jika persamaan 2 punya koefisien x tidak nol
        # tukar seluruh koefisien persamaan 1 dengan persamaan 2
        a1, b1, c1, d1, a2, b2, c2, d2 = a2, b2, c2, d2, a1, b1, c1, d1
    elif a3 != 0:                        # jika persamaan 3 punya koefisien x tidak nol
        # tukar seluruh koefisien persamaan 1 dengan persamaan 3
        a1, b1, c1, d1, a3, b3, c3, d3 = a3, b3, c3, d3, a1, b1, c1, d1
    else:
        # semua koefisien x nol, x tidak bisa dinyatakan dari persamaan mana pun
        raise ValueError("Koefisien x semuanya nol.")

# Menampilkan a1 untuk memastikan nilainya tidak nol
print("a1 =", a1)
```

```python
# ===== CELL 3: SUBSTITUSI x KE PERSAMAAN 2 DAN 3 =====
# Dari persamaan 1 diperoleh: x = (d1 - b1*y - c1*z) / a1
# Rumus x ini dimasukkan ke persamaan 2 dan 3, sehingga x hilang
# dan tersisa dua persamaan dengan variabel y dan z saja.

# Hasil substitusi ke persamaan 2, bentuknya: p1*y + q1*z = r1
p1 = b2 - a2 * b1 / a1      # koefisien y yang baru
q1 = c2 - a2 * c1 / a1      # koefisien z yang baru
r1 = d2 - a2 * d1 / a1      # konstanta yang baru

# Hasil substitusi ke persamaan 3, bentuknya: p2*y + q2*z = r2
p2 = b3 - a3 * b1 / a1      # koefisien y yang baru
q2 = c3 - a3 * c1 / a1      # koefisien z yang baru
r2 = d3 - a3 * d1 / a1      # konstanta yang baru

# Menampilkan dua persamaan baru (hanya berisi y dan z)
print(f"({p1})y + ({q1})z = {r1}")
print(f"({p2})y + ({q2})z = {r2}")
```

```python
# ===== CELL 4: MEMASTIKAN KOEFISIEN y PADA PERSAMAAN BARU 1 TIDAK NOL =====
# Langkah berikutnya menyatakan y dari persamaan baru 1 dengan membagi p1,
# sehingga p1 tidak boleh nol.

if p1 == 0:
    # jika p1 nol, tukar persamaan baru 1 dengan persamaan baru 2
    p1, q1, r1, p2, q2, r2 = p2, q2, r2, p1, q1, r1

if p1 == 0:
    # jika setelah ditukar p1 masih nol, koefisien y nol di kedua persamaan
    # sehingga sistem tidak memiliki solusi tunggal
    raise ValueError("Tidak ada solusi tunggal.")
```

```python
# ===== CELL 5: MENGHITUNG z, y, DAN x =====
# Dari persamaan baru 1: y = (r1 - q1*z) / p1
# Rumus y ini disubstitusikan ke persamaan baru 2, sehingga tersisa satu variabel (z).

# Penyebut dari rumus z, berasal dari hasil substitusi y ke persamaan baru 2
penyebut = q2 - p2 * q1 / p1

# Jika penyebut nol, z tidak bisa dihitung (sistem tidak punya solusi tunggal)
if penyebut == 0:
    raise ValueError("Tidak ada solusi tunggal.")

# Menghitung z dari persamaan yang hanya memuat z
z = (r2 - p2 * r1 / p1) / penyebut

# Menyubstitusikan nilai z ke rumus y untuk mendapatkan y
y = (r1 - q1 * z) / p1

# Menyubstitusikan nilai y dan z ke rumus x untuk mendapatkan x
x = (d1 - b1 * y - c1 * z) / a1
```

```python
# ===== CELL 6: MENAMPILKAN HASIL =====
# f"..." adalah f-string, {x:.4f} berarti nilai x ditampilkan dengan 4 angka di belakang koma
print(f"x = {x:.4f}")
print(f"y = {y:.4f}")
print(f"z = {z:.4f}")
```

```python
# ===== CELL 7: VERIFIKASI HASIL =====
# Memasukkan x, y, z ke ruas kiri tiap persamaan awal.
# Jika benar, hasilnya sama dengan konstanta di ruas kanan (d1, d2, d3).

print(a1*x + b1*y + c1*z, "=", d1)    # persamaan 1: hasil harus sama dengan d1
print(a2*x + b2*y + c2*z, "=", d2)    # persamaan 2: hasil harus sama dengan d2
print(a3*x + b3*y + c3*z, "=", d3)    # persamaan 3: hasil harus sama dengan d3
```
